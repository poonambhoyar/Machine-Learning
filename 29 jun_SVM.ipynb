```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR
```

```python
# generate synthetic data
X = np.sort(5 * np.random.rand(40, 1), axis=0)
y = np.sin(X).ravel()

# ravel():- change a 2-dimensional array or a multi-dimensional array into a contiguous 
```

```python
X
```

```python
X.shape
```

```python
y
```

```python
# This will effectively add noise to every  fifth element in your target array
y[::5] += 3 * (0.5 - np.random.rand(8))
```

```python
y
```

```python
# create an SVR model with a linear kernel
svr = SVR(kernel='linear')
```

```python
# train the model on the data
svr.fit(X, y)
```

```python
# make predictions on the data
y_pred = svr.predict(X)
```

```python
# plat the predicted values against the true values
plt.scatter(X, y, color='darkorange', label='data')
plt.plot(X, y_pred, color='cornflowerblue', label='prediction')
plt.legend()
plt.show()
```
