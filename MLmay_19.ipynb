Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.linspace(0,5,11)
```

```python
a
```

```python
b = a**2
```

```python
b
```

```python
c = np.array([3,5,7,9,10,3])
```

```python
c
```

```python
plt.plot(a,b)
plt.title('SparrowAI',fontsize=20)
plt.xlabel('A',fontsize=18)
plt.ylabel('B',fontsize=18)
plt.show()
```

```python
plt.scatter(a,b)
plt.title('SparrowAI',fontsize=20)
plt.xlabel('A',fontsize=18)
plt.ylabel('B',fontsize=18)
plt.show()
```

```python
plt.bar(a,b)
plt.title('SparrowAI',fontsize=20)
plt.xlabel('A',fontsize=18)
plt.ylabel('B',fontsize=18)
plt.show()
```

```python
plt.pie(c)
```

Subplot

```python
plt.subplot(1,3,1)    # width, height, subplot no
plt.plot(a,b)

plt.subplot(1,3,2)
plt.plot(b,a)

plt.subplot(1,3,3)
plt.plot(a,a)

plt.tight_layout()
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.6,0.6])
axes.plot(a,b)
axes.set_title('SparrowAI')
axes.set_xlabel('A')
axes.set_ylabel('B')
plt.show()
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig.add_axes([0.2,0.5,0.4,0.3])

axes1.plot(a,b)
axes2.plot(b,a)

axes.set_title('SparrowAI')
axes.set_xlabel('A')
axes.set_ylabel('B')
plt.show()
```
