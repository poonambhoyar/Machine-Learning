```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('darkgrid')
```

```python
x = [2,3,4,6,5,8,9]
```

```python
x
```

```python
y = [5,7,9,8,2,6,10]
```

```python
y
```

```python
plt.scatter(x,y)
plt.show()
```

```python
fig,axes = plt.subplots(nrows=2,ncols=2)
plt.tight_layout()
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)
axes[0].plot(x,y)
axes[1].plot(y,x)
plt.tight_layout()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y, label='test1')
ax.plot(y,x, label='test2')
plt.legend()
plt.show()
```
