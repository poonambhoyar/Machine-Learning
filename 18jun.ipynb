```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('darkgrid')
```

```python
data = pd.read_csv('Position_Salaries.csv')
```

```python
data
```

```python
x = data.iloc[:,1].values
```

```python
x
```

```python
y = data.iloc[:,2].values
```

```python
y
```

```python
plt.scatter(x,y,c='g')
plt.xlabel('level')
plt.ylabel('salary')
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
reg = DecisionTreeRegressor()
```

```python
reg.fit(x.reshape(-1,1),y)
```

```python
plt.scatter(x,y,c='g')
plt.plot(x,reg.predict(x.reshape(-1,1)),c='b')
plt.xlabel('level')
plt.ylabel('salary')
plt.show()
```
