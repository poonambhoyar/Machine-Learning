```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np
```

```python
# creating datasets X containing n_samples

# Y containing two classes

X, Y = make_blobs(n_samples=500, centers=2,random_state=0, cluster_std=0.40)  #cluster_std
```

```python
X.shape
```

```python
X
```

```python
# plotting scatte

plt.scatter(X[:, 0], X[:, 1], c=Y, s=50, cmap='spring')  #s data point size
plt.show()
```

```python
# creating linspace between -1 to 3.5
xfit = np.linspace(-1, 3.5)
```

```python
xfit.shape
```

```python
xfit
```

```python
# plotting scatter

plt.scatter(X[:, 0], X[:, 1], c=Y, s=50, cmap='spring')

# plot a line between the different sets of data
# 1st value: width of line
# 2nd value: height of line
# 3rd value: filled boundries of margine line

for m, b, d in [(1, 0.65,0.33), (-0.2, 2.9, 0.2), (0.5,1.6, 0.55)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit,'-k')   # -k = not alloting colors to line
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none', color='#AAAAAA', alpha=0)
plt.xlim(-1, 3.5);
plt.show()
```
